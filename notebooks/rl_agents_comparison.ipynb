## Consolidating data and plotting the curves together

```python
import os
import random
import pandas as pd
import numpy as np
from loguru import logger

from pandemic_control.utils import (
    Args,
    set_plot_env,
    plot_rewards,
)
```

```python
args = Args(
    const_policy_simulations=False,
    model_policy_simulations= False,
    plot_heatmaps=False,
    plot_learning_curves=False,
    preprocess_data=False,
    train=True,
    predict=False,
    env_type='SEIRADHV',
    cfg_file='./configs/envs_tests/default-seiradhv.json',
    model_type='PPO',
    model_weights=None,
    output_dir='./outputs',
    seed=random.choice([33, 45, 75, 99]),
    t_max=730,
    timesteps=832,
    epochs=100,
    save_interval=250,
    )



model_types = ['PPO', 'A2C', 'QRDQN', 'DQN', 'DDQN']
rewards_per_step = None
rewards_per_episodes = None
for model_type in model_types:
    args.model_type = model_type
    output_dir = os.path.join(
        f"{args.output_dir}", 
        f"learning_curves",
        f"{args.env_type.lower()}_{args.model_type.lower()}", 
        "history"
        )
    df = pd.read_csv(os.path.join(f"{output_dir}", f"reward_per_step.csv"))
    print(args.model_type, 'Number of steps: ', len(df))
    if rewards_per_step is None:
        rewards_per_step = df
    else:
        if len(df) > len(rewards_per_step):
            rewards_per_step = pd.concat([rewards_per_step, df[[f'{args.model_type}']].iloc[:len(rewards_per_step)]], axis=1)
        elif len(df) < len(rewards_per_step):
            rewards_per_step = pd.concat([rewards_per_step.iloc[:len(df)], df[[f'{args.model_type}']]], axis=1)
        else:
            rewards_per_step = pd.concat([rewards_per_step, df[[f'{args.model_type}']]], axis=1)

    df = pd.read_csv(os.path.join(f"{output_dir}", f"reward_per_episode.csv"))
    print(args.model_type, 'Number of episodes: ', len(df))
    if rewards_per_episodes is None:
        rewards_per_episodes = df
    else:
        if len(df) > len(rewards_per_episodes):
            rewards_per_episodes = pd.concat([rewards_per_episodes, df[[f'{args.model_type}']].iloc[:len(rewards_per_episodes)]], axis=1)
        elif len(df) < len(rewards_per_episodes):
            rewards_per_episodes = pd.concat([rewards_per_episodes.iloc[:len(df)], df[[f'{args.model_type}']]], axis=1)
        else:
            rewards_per_episodes = pd.concat([rewards_per_episodes, df[[f'{args.model_type}']]], axis=1)
```
